```python
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Dense, Conv1D, MaxPooling1D, Flatten, GlobalAveragePooling1D, BatchNormalization, Activation, Add, Input, Dropout
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
def create_vggnet_1d(input_shape, num_classes):
    model = Sequential()

    # Block 1
    model.add(Conv1D(32, 3, activation='relu', padding='same', input_shape=input_shape))
    model.add(BatchNormalization())
    model.add(Conv1D(32, 3, activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling1D(2, padding='same'))
    model.add(Dropout(0.2))

    # Block 2
    model.add(Conv1D(64, 3, activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv1D(64, 3, activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling1D(2, padding='same'))
    model.add(Dropout(0.3))

    # Block 3
    model.add(Conv1D(128, 3, activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(Conv1D(128, 3, activation='relu', padding='same'))
    model.add(BatchNormalization())
    model.add(MaxPooling1D(2, padding='same'))
    model.add(Dropout(0.4))

    # Classification head
    model.add(Flatten())
    model.add(Dense(256, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.001)))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    model.add(Dense(128, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.001)))
    model.add(BatchNormalization())
    model.add(Dropout(0.5))
    model.add(Dense(num_classes, activation='softmax'))

    return model
```

```python
def residual_block(x, filters, kernel_size=3, stride=1, dropout_rate=0.2):
    shortcut = x
    x = Conv1D(filters, kernel_size, strides=stride, padding='same')(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Dropout(dropout_rate)(x)

    x = Conv1D(filters, kernel_size, padding='same')(x)
    x = BatchNormalization()(x)

    if stride != 1 or shortcut.shape[-1] != filters:
        shortcut = Conv1D(filters, 1, strides=stride, padding='same')(shortcut)
        shortcut = BatchNormalization()(shortcut)

    x = Add()([x, shortcut])
    x = Activation('relu')(x)
    x = Dropout(dropout_rate)(x)

    return x

def create_resnet_1d(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    x = Conv1D(64, 7, strides=2, padding='same')(inputs)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = MaxPooling1D(3, strides=2, padding='same')(x)
    x = Dropout(0.2)(x)

    x = residual_block(x, 64, dropout_rate=0.2)
    x = residual_block(x, 64, dropout_rate=0.2)
    x = residual_block(x, 128, stride=2, dropout_rate=0.3)
    x = residual_block(x, 128, dropout_rate=0.3)
    x = residual_block(x, 256, stride=2, dropout_rate=0.4)
    x = residual_block(x, 256, dropout_rate=0.4)

    x = GlobalAveragePooling1D()(x)
    x = Dense(512, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.001))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    x = Dense(256, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.001))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.5)(x)
    outputs = Dense(num_classes, activation='softmax')(x)

    return Model(inputs, outputs)
```

```python
def plot_epoch_comparison(model_results, model_name, target_name):
    epochs = [r['Epochs'] for r in model_results]
    train_acc = [r['Train Accuracy'] for r in model_results]
    test_acc = [r['Test Accuracy'] for r in model_results]
    train_loss = [r['Train Loss'] for r in model_results]
    test_loss = [r['Test Loss'] for r in model_results]

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    ax1.plot(epochs, train_acc, 'b-o', label='Train Accuracy')
    ax1.plot(epochs, test_acc, 'r-s', label='Test Accuracy')
    ax1.set_title(f'{model_name} Accuracy vs Epochs\n{target_name}')
    ax1.legend(); ax1.grid(True)

    ax2.plot(epochs, train_loss, 'b-o', label='Train Loss')
    ax2.plot(epochs, test_loss, 'r-s', label='Test Loss')
    ax2.set_title(f'{model_name} Loss vs Epochs\n{target_name}')
    ax2.legend(); ax2.grid(True)

    plt.tight_layout()
    plt.show()

def plot_training_history(history, model_name, epochs, target_name):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    ax1.plot(history.history['accuracy'], label='Train')
    ax1.plot(history.history['val_accuracy'], label='Validation')
    ax1.set_title(f'{model_name} Accuracy ({epochs} epochs)\n{target_name}')
    ax1.legend(); ax1.grid(True)

    ax2.plot(history.history['loss'], label='Train')
    ax2.plot(history.history['val_loss'], label='Validation')
    ax2.set_title(f'{model_name} Loss ({epochs} epochs)\n{target_name}')
    ax2.legend(); ax2.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
# ===== Cell 5 (REPLACEMENT) — Core training function (filters only the columns you specified) =====
def train_deep_learning_models_iterative(X_df, Y_df, target_name):
    """
    X_df: pandas DataFrame of features (this may be the df_train.drop([...]) in your flow)
    Y_df: single-column pandas DataFrame containing the target
    target_name: str, column name of the target (for printing/plots)

    This function restricts features to a predefined allowed list (the columns you provided),
    drops identifier/name columns automatically (to avoid parsing issues), performs dummy encoding,
    scaling, reshaping, then trains VGGNet and ResNet models for multiple epoch settings.
    """
    print(f"\n{'='*60}")
    print(f"Training Deep Learning Models for: {target_name}")
    print(f"{'='*60}")

    # --- Allowed columns (only these will be considered) ---
    allowed_columns = [
        "Patient Id", "Patient Age", "Genes in mother's side", "Inherited from father",
        "Maternal gene", "Paternal gene", "Blood cell count (mcL)",
        "Patient First Name", "Family Name", "Father's name",
        # ... if you have more columns between names and Birth defects add them here ...
        "Birth defects", "White Blood cell count (thousand per microliter)",
        "Blood test result", "Symptom 1", "Symptom 2", "Symptom 3",
        "Symptom 4", "Symptom 5", "Genetic Disorder", "Disorder Subclass"
    ]

    # Ensure target is present
    if target_name not in Y_df.columns and target_name in X_df.columns:
        # If the user passed combined df with target in X_df, move it
        Y_df = X_df[[target_name]].copy()
        X_df = X_df.drop(columns=[target_name])

    # Intersect allowed with actual columns present
    present_allowed = [c for c in allowed_columns if c in X_df.columns or c in Y_df.columns]
    if len(present_allowed) == 0:
        raise ValueError("None of the allowed columns were found in the input dataframe. Check column names.")

    # Select features from X_df (but avoid selecting the target column if it's listed)
    features_to_select = [c for c in present_allowed if c != target_name and c in X_df.columns]
    print(f"Features present and selected (pre-drop): {features_to_select}")

    # Automatically drop identifier/name columns because they often cause parsing/high-cardinality issues.
    identifier_like = [c for c in features_to_select if ("Id" in c or "Name" in c or "First Name" in c or "Family Name" in c or "Father's name" in c)]
    if identifier_like:
        print(f"Dropping identifier/name-like columns to avoid parsing issues: {identifier_like}")
        features_to_select = [c for c in features_to_select if c not in identifier_like]

    if len(features_to_select) == 0:
        raise ValueError("No usable features remain after dropping identifier/name-like columns. "
                         "If you want to keep them, modify the function to allow them.")

    X_sel = X_df[features_to_select].copy()
    Y_sel = Y_df[target_name].copy() if target_name in Y_df.columns else Y_df.iloc[:, 0].copy()

    # --- Diagnostics: show dtypes and a small sample ---
    print("\nSelected feature dtypes:\n", X_sel.dtypes)
    print("\nSample of selected features:\n", X_sel.head(3))

    # --- Handle missing values: simple imputation (numerical -> median, categorical -> mode) ---
    for col in X_sel.columns:
        if X_sel[col].dtype.kind in 'biufc':  # numeric
            if X_sel[col].isna().any():
                med = X_sel[col].median()
                X_sel[col].fillna(med, inplace=True)
                print(f"Filled NA in numeric column '{col}' with median: {med}")
        else:
            if X_sel[col].isna().any():
                mode_val = X_sel[col].mode().iloc[0] if not X_sel[col].mode().empty else "missing"
                X_sel[col].fillna(mode_val, inplace=True)
                print(f"Filled NA in categorical column '{col}' with mode: {mode_val}")

    # --- Convert categorical columns to numeric via one-hot / dummy encoding ---
    cat_cols = X_sel.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
    if cat_cols:
        print(f"One-hot encoding categorical columns: {cat_cols}")
        X_sel = pd.get_dummies(X_sel, columns=cat_cols, drop_first=True)
    else:
        print("No categorical columns to one-hot encode.")

    # Make sure all columns are numeric now
    non_numeric_after = X_sel.select_dtypes(exclude=[np.number]).columns.tolist()
    if non_numeric_after:
        # try to coerce
        print(f"Attempting to coerce remaining non-numeric columns to numeric: {non_numeric_after}")
        for c in non_numeric_after:
            X_sel[c] = pd.to_numeric(X_sel[c], errors='coerce')
            if X_sel[c].isna().any():
                X_sel[c].fillna(X_sel[c].median(), inplace=True)

    # Final check
    if X_sel.isna().any().any():
        raise ValueError("There are still NaNs in the feature matrix after imputation. Inspect the data.")

    # --- Encode target ---
    le = LabelEncoder()
    Y_encoded = le.fit_transform(Y_sel.values.ravel())
    num_classes = len(np.unique(Y_encoded))
    Y_categorical = to_categorical(Y_encoded, num_classes=num_classes)
    print(f"Target '{target_name}' encoded into {num_classes} classes: {le.classes_}")

    # --- Train/test split ---
    x_train, x_test, y_train, y_test = train_test_split(
        X_sel.values, Y_categorical, test_size=0.2, random_state=42, stratify=Y_encoded
    )

    # --- Scale numeric features ---
    scaler = StandardScaler()
    x_train_scaled = scaler.fit_transform(x_train)
    x_test_scaled = scaler.transform(x_test)

    # --- Reshape for 1D CNN ---
    x_train_reshaped = x_train_scaled.reshape(x_train_scaled.shape[0], x_train_scaled.shape[1], 1)
    x_test_reshaped = x_test_scaled.reshape(x_test_scaled.shape[0], x_test_scaled.shape[1], 1)
    input_shape = (x_train_scaled.shape[1], 1)
    print(f"Final input shape for models: {input_shape}")

    # --- Models to train ---
    models = {
        'VGGNet': create_vggnet_1d(input_shape, num_classes),
        'ResNet': create_resnet_1d(input_shape, num_classes)
    }

    # Epoch configurations (you can change)
    epoch_configs = [50, 100]
    batch_size = 64
    learning_rate = 0.001

    all_results = {}

    for model_name in models.keys():
        print(f"\n{'-'*40}\nTraining model: {model_name}\n{'-'*40}")
        model_results = []

        for epochs in epoch_configs:
            print(f"\nTraining {model_name} with {epochs} epochs...")
            # recreate model for each run
            current_model = (create_vggnet_1d if model_name == 'VGGNet' else create_resnet_1d)(input_shape, num_classes)
            current_model.compile(
                optimizer=Adam(learning_rate=learning_rate, beta_1=0.9, beta_2=0.999, epsilon=1e-7),
                loss='categorical_crossentropy',
                metrics=['accuracy']
            )

            callbacks = [
                EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True, verbose=0),
                ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=10, verbose=0)
            ]

            history = current_model.fit(
                x_train_reshaped, y_train,
                batch_size=batch_size,
                epochs=epochs,
                validation_data=(x_test_reshaped, y_test),
                verbose=1,
                callbacks=callbacks,
                shuffle=True
            )

            train_loss, train_accuracy = current_model.evaluate(x_train_reshaped, y_train, verbose=0)
            test_loss, test_accuracy = current_model.evaluate(x_test_reshaped, y_test, verbose=0)
            y_pred_proba = current_model.predict(x_test_reshaped, verbose=0)
            y_pred = np.argmax(y_pred_proba, axis=1)
            y_true = np.argmax(y_test, axis=1)

            print(f"Epochs: {epochs} | Train Acc: {train_accuracy:.4f} | Test Acc: {test_accuracy:.4f}")
            model_results.append({
                'Epochs': epochs,
                'Train Accuracy': train_accuracy,
                'Test Accuracy': test_accuracy,
                'Train Loss': train_loss,
                'Test Loss': test_loss,
                'Model': current_model,
                'History': history,
                'Predictions': y_pred,
                'True Labels': y_true
            })

        all_results[model_name] = model_results

        # Plot epoch comparison and training history for best configuration
        plot_epoch_comparison(model_results, model_name, target_name)
        best_result = max(model_results, key=lambda x: x['Test Accuracy'])
        print(f"\nBest configuration for {model_name}: Epochs={best_result['Epochs']} | Test Acc={best_result['Test Accuracy']:.4f}")
        plot_training_history(best_result['History'], model_name, best_result['Epochs'], target_name)

        # Confusion matrix & report
        cm = confusion_matrix(best_result['True Labels'], best_result['Predictions'])
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=le.classes_, yticklabels=le.classes_)
        plt.title(f'Confusion Matrix - {model_name} (Best: {best_result["Epochs"]} epochs)\nTarget: {target_name}')
        plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.tight_layout(); plt.show()

        print(f"\nClassification Report for {model_name} (Best: {best_result['Epochs']} epochs):")
        print(classification_report(best_result['True Labels'], best_result['Predictions'], target_names=le.classes_))

    return all_results
```

```python
target_columns = ['Genetic Disorder', 'Disorder Subclass', 'Combined_disorder']
all_target_results = {}

for target_col in target_columns:
    if target_col in df_train.columns:
        print(f"\nProcessing Target: {target_col}")
        X = df_train.drop(columns=[target_col])
        Y = df_train[[target_col]]
        results = train_deep_learning_models_iterative(X, Y, target_col)
        all_target_results[target_col] = results
```

```python
summary_data = []
for target_col, model_results in all_target_results.items():
    for model_name, epochs_results in model_results.items():
        best_epoch = max(epochs_results, key=lambda x: x['Test Accuracy'])
        summary_data.append({
            'Target': target_col, 'Model': model_name,
            'Best Epochs': best_epoch['Epochs'],
            'Best Test Accuracy': f"{best_epoch['Test Accuracy']:.4f}",
            'Best Train Accuracy': f"{best_epoch['Train Accuracy']:.4f}"
        })

summary_df = pd.DataFrame(summary_data)
display(summary_df)
```
